```python
df = pd.DataFrame(np.random.randn(100,4),columns='A B C D'.split())
```

```python
df2 = pd.DataFrame({'Category':['A','B','C'],'Values':[32,43,50]})
```

```python
import plotly.io as pio
pio.renderers.default = "notebook_connected"
```

```python
import plotly.express as px

fig = px.scatter(df, x='A', y='B', size=None, color_discrete_sequence=['#ff9933'])
fig.show()
```

```python
fig = px.bar(df2, x="Category", y="Values")
fig.show()
```

```python
fig = px.bar(x=df.columns, y=df.count().values)
fig.update_layout(xaxis_title="Columns", yaxis_title="Count")
fig.show()
```

```python
fig = px.box(df, y=df.columns)  # shows all columns
fig.show()
```

```python
fig = px.line(df, x=df.index, y=["A", "B"])
fig.show()
```

```python
fig = px.histogram(df, x="A", nbins=25)
fig.show()
```

```python
fig = px.scatter_matrix(df)
fig.show()
```
