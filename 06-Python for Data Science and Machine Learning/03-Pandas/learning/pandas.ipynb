```python
import pandas as pd
import numpy as np

# reproducibility
np.random.seed(42)

# create sample dataset
data = pd.DataFrame({
    "CustomerID": np.random.randint(1001, 1010, size=20),
    "Region": np.random.choice(["North", "South", "East", "West"], size=20),
    "Product": np.random.choice(["Laptop", "Mobile", "Tablet", "Headphones"], size=20),
    "Quantity": np.random.randint(1, 5, size=20),
    "Price": np.random.randint(100, 1000, size=20),
    "PurchaseDate": pd.date_range("2025-01-01", periods=20, freq="D")
})

print(data)
```

```python
data.groupby('Product')['Price'].sum()
```

```python
print(data.groupby('Region')['Quantity'].sum().min())
```

```python
data.groupby('CustomerID').describe().transpose()
```
