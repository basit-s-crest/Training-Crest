```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
```

```python
x = np.linspace(0,5,11)
y = x ** 2
```

```python
plt.plot(x,y,'-y')
plt.show()
```

```python
plt.plot(x,y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Title')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'-r')
plt.show()

plt.subplot(2,1,1)
plt.plot(x,y,'-g')
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes([1,1.5,2,2.5])
axes.plot(x,y)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([1,2,3,4])
axis2 = fig.add_axes([0.1,0.2,0.3,0.4])

axis1.plot(x,y)
axis2.plot(x,y)
```

```python
fig , axes = plt.subplots(nrows=3,ncols=3)
#plt.show()
```

```python
x = [i ** 2 for i in range(5)]
y = [i for i in range(5)]
```

```python
plt.plot(y,x,lw=2,ls='-',marker='o')
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(y,x,color='orange',lw=2,ls='-',marker='o')
ax.set_xlim([1,2])
```
